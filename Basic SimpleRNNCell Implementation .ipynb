```python
import numpy as np

class SimpleRNNCell:
    def __init__(self, input_size, hidden_size):
        # Initialize weights with small random values
        self.W_xh = np.random.randn(hidden_size, input_size) * 0.01  # Input to hidden
        self.W_hh = np.random.randn(hidden_size, hidden_size) * 0.01 # Hidden to hidden
        self.b_h = np.zeros((hidden_size, 1))                        # Hidden bias

    def forward(self, x_t, h_prev):
        """
        x_t: input vector at time t, shape (input_size, 1)
        h_prev: previous hidden state, shape (hidden_size, 1)
        """
        # Calculate new hidden state
        h_t = np.tanh(np.dot(self.W_xh, x_t) + np.dot(self.W_hh, h_prev) + self.b_h)
        return h_t

# --- Example Usage ---
input_size = 4   # e.g., size of vocabulary or feature vector
hidden_size = 5  # size of hidden state representation
seq_length = 3   # number of time steps

rnn_cell = SimpleRNNCell(input_size, hidden_size)

# Dummy sequence of inputs (3 time steps, batch of 1)
inputs = [np.random.randn(input_size, 1) for _ in range(seq_length)]
h_t = np.zeros((hidden_size, 1))  # Initial hidden state (h_0)

print("Forward pass across sequence:")
for t, x_t in enumerate(inputs):
    h_t = rnn_cell.forward(x_t, h_t)
    print(f"Time Step {t+1} Hidden State Shape:", h_t.shape)
```
