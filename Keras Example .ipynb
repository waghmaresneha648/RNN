```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# 1. Create dummy input data: 5 samples, sequence length 3, 1 feature per step
X = np.random.randn(5, 3, 1)

# 2. Build the model
model = Sequential([
    SimpleRNN(units=4, input_shape=(3, 1)), # Simple RNN layer with 4 hidden neurons
    Dense(units=1)                           # Output layer for prediction
])

# 3. Compile and run a single forward prediction
model.compile(optimizer='adam', loss='mse')
predictions = model.predict(X)

print("Predictions:\n", predictions)
```
